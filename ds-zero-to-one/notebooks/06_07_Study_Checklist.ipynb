```python
"""
Generate Study Session checklist PDFs for:

- Session 6: Residual Diagnostics & Heteroscedasticity
- Session 7: Multicollinearity & VIF

Outputs:
- Study_Session_6_Checklist.pdf
- Study_Session_7_Checklist.pdf
"""

from reportlab.lib.pagesizes import letter
from reportlab.pdfgen import canvas
from reportlab.lib.units import inch

def create_checklist_pdf(filename, title, subtitle, sections):
    """
    Create a simple checklist-style PDF.

    filename : str
        Output PDF filename.
    title : str
        Main title at the top of the page.
    subtitle : str
        Short description or session focus.
    sections : list of dicts
        Each dict: {"heading": str, "items": [str, ...]}
    """
    c = canvas.Canvas(filename, pagesize=letter)
    width, height = letter

    # Margins and line spacing
    left_margin = 0.9 * inch
    right_margin = 0.9 * inch
    top_margin = height - 1.0 * inch
    bottom_margin = 0.9 * inch
    line_height = 14  # points

    y = top_margin

    # --- Title ---
    c.setFont("Helvetica-Bold", 18)
    c.drawString(left_margin, y, title)
    y -= 1.2 * line_height

    # --- Subtitle ---
    if subtitle:
        c.setFont("Helvetica", 11)
        c.drawString(left_margin, y, subtitle)
        y -= 1.5 * line_height

    # --- Sections ---
    for section in sections:
        heading = section.get("heading", "")
        items = section.get("items", [])

        # Add heading
        if y < bottom_margin + 3 * line_height:
            c.showPage()
            y = top_margin
            c.setFont("Helvetica-Bold", 18)
            c.drawString(left_margin, y, title + " (cont.)")
            y -= 1.5 * line_height

        c.setFont("Helvetica-Bold", 12)
        c.drawString(left_margin, y, heading)
        y -= 1.1 * line_height

        # Add items with checkboxes
        c.setFont("Helvetica", 11)
        for item in items:
            if y < bottom_margin + 2 * line_height:
                c.showPage()
                y = top_margin
                c.setFont("Helvetica-Bold", 18)
                c.drawString(left_margin, y, title + " (cont.)")
                y -= 1.5 * line_height
                c.setFont("Helvetica", 11)

            # Draw checkbox
            box_size = 10
            c.rect(left_margin, y - box_size + 3, box_size, box_size, stroke=1, fill=0)

            # Draw text to the right of checkbox
            text_x = left_margin + box_size + 6
            c.drawString(text_x, y, item)
            y -= line_height

        y -= 0.5 * line_height  # extra space between sections

    c.showPage()
    c.save()
    print(f"✅ Created: {filename}")


# =========================
# Session 6 Checklist Data
# =========================

session6_title = "Study Session 6 Checklist"
session6_subtitle = "Residual Diagnostics, Heteroscedasticity, and Log-Transformed Models"

session6_sections = [
    {
        "heading": "Setup & Data",
        "items": [
            "Open the notebook for Session 6 (Residual Diagnostics / Heteroscedasticity).",
            "Load data/processed/tips_cleaned.csv into a DataFrame df.",
            "Confirm numeric columns: bill_total_usd, tip_usd, party_size, tip_pct.",
        ],
    },
    {
        "heading": "Baseline Linear Model",
        "items": [
            "Define X = [bill_total_usd, party_size] and y = tip_usd.",
            "Fit a baseline LinearRegression (scikit-learn) or OLS (statsmodels).",
            "Store predictions y_pred and residuals = y - y_pred.",
            "Compute baseline metrics (RMSE, R²) and record them in a Markdown cell.",
        ],
    },
    {
        "heading": "Residual vs Predicted Plot",
        "items": [
            "Create a scatter plot: x = predicted tips, y = residuals.",
            "Add horizontal line at 0 (perfect predictions).",
            "Check visually: Are residuals randomly scattered (good)?",
            "Look for funnel shape (increasing spread) → possible heteroscedasticity.",
            "Save figure using save_and_log_plot (if available).",
        ],
    },
    {
        "heading": "Residual Distribution & Normality",
        "items": [
            "Plot histogram + KDE of residuals (sns.histplot with kde=True).",
            "Create a Q–Q plot using scipy.stats.probplot or statsmodels API.",
            "Check visually: Do residuals roughly follow a straight line in Q–Q?",
            "Note deviations in the tails or skewness in a Markdown cell.",
        ],
    },
    {
        "heading": "Homoscedasticity (Equal-Variance) Checks",
        "items": [
            "Re-examine residuals vs predicted for patterns in variance.",
            "Optionally run Breusch–Pagan test using statsmodels.stats.diagnostic.het_breuschpagan.",
            "Record BP test statistic and p-value in a Markdown cell.",
            "Interpret result: p < 0.05 suggests heteroscedasticity.",
        ],
    },
    {
        "heading": "Log-Transformed Model (Log Tip)",
        "items": [
            "Create y_log = np.log1p(tip_usd) to stabilize variance.",
            "Fit new model: y_log ~ bill_total_usd + party_size.",
            "Generate predictions on log scale and back-transform via np.expm1.",
            "Compute RMSE and R² for the log model (in original tip_usd units).",
            "Compare baseline vs log-model metrics in a small table.",
        ],
    },
    {
        "heading": "Residual Diagnostics for Log Model",
        "items": [
            "Plot residuals vs predicted for the log-based model.",
            "Plot histogram + Q–Q plot of new residuals.",
            "Check if variance is more constant across predicted values.",
            "Summarize improvements (or not) in a Markdown ELI5 explanation.",
        ],
    },
    {
        "heading": "Reflection & Storytelling",
        "items": [
            "Write a short paragraph explaining what heteroscedasticity is.",
            "Explain why log-transforming tip_usd can help stabilize variance.",
            "Describe which model you would report to a stakeholder and why.",
        ],
    },
]

# =========================
# Session 7 Checklist Data
# =========================

session7_title = "Study Session 7 Checklist"
session7_subtitle = "Multicollinearity and VIF (Variance Inflation Factor)"

session7_sections = [
    {
        "heading": "Setup & Encoded Features",
        "items": [
            "Open the notebook for Session 7 (Multicollinearity & VIF).",
            "Load data/processed/tips_cleaned.csv into df.",
            "Create df_encoded = df.copy().",
            "Create gender_Male as 1 if gender is male (m/M), else 0.",
            "Create is_smoker_Yes as 1 if is_smoker is yes (y/true/1), else 0.",
            "Confirm dtypes of gender_Male and is_smoker_Yes are numeric (int).",
        ],
    },
    {
        "heading": "Select Features for VIF",
        "items": [
            "Choose features = [bill_total_usd, party_size, gender_Male, is_smoker_Yes].",
            "Check for any missing or non-numeric values in these columns.",
            "Optionally drop rows with missing values in the chosen features.",
        ],
    },
    {
        "heading": "Implement calculate_vif Helper",
        "items": [
            "Define calculate_vif(df, features) using statsmodels' variance_inflation_factor.",
            "Inside the function: build X = df[features].copy().",
            "Add a constant column using statsmodels.tools.add_constant.",
            "Compute VIF for each column and return a neat DataFrame.",
        ],
    },
    {
        "heading": "Compute and Inspect VIF Values",
        "items": [
            "Call vif_table = calculate_vif(df_encoded, features).",
            "Display the table and confirm VIF values for each feature.",
            "Check if any VIF > 5 (moderate) or > 10 (severe).",
            "Record your interpretation in Markdown (e.g., no serious issues here).",
        ],
    },
    {
        "heading": "VIF Visualization",
        "items": [
            "Create a bar chart of VIF vs Feature using seaborn or matplotlib.",
            "Add horizontal reference lines at VIF=5 and VIF=10 (traffic-light thresholds).",
            "Color-code bars (e.g., green <5, yellow 5–10, red >10).",
            "Save figure using save_and_log_plot (if available).",
        ],
    },
    {
        "heading": "Interpretation & Domain Insight",
        "items": [
            "Explain why bill_total_usd and party_size have some correlation (but acceptable VIF).",
            "Explain why gender_Male and is_smoker_Yes have VIF near 1 (independent).",
            "Describe in plain language what multicollinearity does to regression coefficients.",
            "Note: High VIF = unstable, noisy, hard-to-interpret coefficients.",
        ],
    },
    {
        "heading": "Fixing Multicollinearity (Hypothetical Scenarios)",
        "items": [
            "Write down at least 3 ways to handle high VIF (e.g., drop, combine, regularize).",
            "Give an example of creating a new feature like bill_per_person.",
            "Explain when you might use Ridge regression to stabilize coefficients.",
        ],
    },
    {
        "heading": "Reusable Diagnostics",
        "items": [
            "Ensure calculate_vif(df, features) is saved in a helper module or a reusable cell.",
            "Add a Markdown note with the VIF interpretation table (1, 5, 10 thresholds).",
            "Optionally build a small function that prints a VIF warning if any feature > 10.",
        ],
    },
]

# =========================
# Generate both PDFs
# =========================

create_checklist_pdf(
    "Study_Session_6_Checklist.pdf",
    session6_title,
    session6_subtitle,
    session6_sections,
)

create_checklist_pdf(
    "Study_Session_7_Checklist.pdf",
    session7_title,
    session7_subtitle,
    session7_sections,
)
```
