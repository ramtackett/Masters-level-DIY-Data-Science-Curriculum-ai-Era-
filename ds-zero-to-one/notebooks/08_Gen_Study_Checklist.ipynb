```python
from reportlab.lib.pagesizes import letter
from reportlab.lib.units import inch
from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer, ListFlowable, ListItem
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib.enums import TA_CENTER

# --------- Config ---------
OUTPUT_PATH = CORRECT_PROJECT_ROOT + "/../PDF/Study_Session_8_Checklist.pdf"  # or "reports/Study_Session_8_Checklist.pdf"

# --------- Styles ---------
styles = getSampleStyleSheet()

title_style = ParagraphStyle(
    "TitleStyle",
    parent=styles["Title"],
    alignment=TA_CENTER,
    fontSize=22,
    leading=26,
    spaceAfter=14,
)

subtitle_style = ParagraphStyle(
    "SubtitleStyle",
    parent=styles["Heading2"],
    alignment=TA_CENTER,
    fontSize=12,
    leading=14,
    textColor="#555555",
    spaceAfter=20,
)

h1 = ParagraphStyle(
    "H1",
    parent=styles["Heading1"],
    fontSize=16,
    leading=18,
    spaceAfter=8,
)

h2 = ParagraphStyle(
    "H2",
    parent=styles["Heading2"],
    fontSize=13,
    leading=15,
    spaceAfter=6,
)

body = ParagraphStyle(
    "Body",
    parent=styles["BodyText"],
    fontSize=10,
    leading=13,
    spaceAfter=6,
)


def bullet_list(items, style=body, left_indent=16):
    """
    Convenience helper to build a checklist-style bullet list.
    Each item should already contain '[ ]' at the start.
    """
    return ListFlowable(
        [ListItem(Paragraph(text, style)) for text in items],
        bulletType="bullet",
        start=None,
        leftIndent=left_indent,
    )


# --------- Build Document ---------
doc = SimpleDocTemplate(
    OUTPUT_PATH,
    pagesize=letter,
    leftMargin=0.75 * inch,
    rightMargin=0.75 * inch,
    topMargin=0.75 * inch,
    bottomMargin=0.75 * inch,
)

story = []

# Title
story.append(Paragraph("Study Session 8 — Non-Linear Models & GAMs", title_style))
story.append(Paragraph("Masters-Level DIY Data Science Curriculum (Tips Dataset)", subtitle_style))
story.append(Spacer(1, 0.15 * inch))

# Section 1: Warm-Up & Context
story.append(Paragraph("1. Warm-Up & Context (5–10 min)", h1))
story.append(bullet_list([
    "[ ] Skim prior sessions: multiple regression, VIF, regularization.",
    "[ ] Remind yourself why linear models can be too rigid (only straight lines).",
    "[ ] Write a one-sentence goal for this session (e.g., “Understand what a GAM buys me over OLS”).",
]))
story.append(Spacer(1, 0.1 * inch))

# Section 2: Load Data & Simple Checks
story.append(Paragraph("2. Load Data & Simple Checks (5–10 min)", h1))
story.append(bullet_list([
    "[ ] Load tips dataset (e.g., tips_cleaned or tips_synthetic) into df.",
    "[ ] Confirm columns: bill_total_usd, party_size, tip_usd, gender, is_smoker, tip_pct.",
    "[ ] Create df_work copy and safely encode gender_Male and is_smoker_Yes if needed.",
    "[ ] Run df_work.describe() and sanity-check ranges.",
]))
story.append(Spacer(1, 0.1 * inch))

# Section 3: Why Non-Linearity & Where Linear Regression Struggles
story.append(Paragraph("3. Non-Linearity & Where Linear Regression Struggles (10–15 min)", h1))
story.append(bullet_list([
    "[ ] Plot tip_usd vs bill_total_usd with a straight regression line (OLS).",
    "[ ] Inspect residuals vs predicted tips; look for curvature or fanning patterns.",
    "[ ] Jot down 1–2 spots where you suspect non-linear behavior (e.g., very large bills, large party_size).",
]))
story.append(Spacer(1, 0.1 * inch))

# Section 4: Fit a Baseline Linear Model (Reference Point)
story.append(Paragraph("4. Baseline Linear Model (Reference Point) (10–15 min)", h1))
story.append(bullet_list([
    "[ ] Fit a multiple linear regression: tip_usd ~ bill_total_usd + party_size + gender_Male + is_smoker_Yes.",
    "[ ] Record R², Adjusted R², and RMSE (train and/or test).",
    "[ ] Save ‘Actual vs Predicted Tip’ plot with save_and_log_plot().",
    "[ ] Save residuals vs predicted plot to compare later against GAM.",
]))
story.append(Spacer(1, 0.1 * inch))

# Section 5: Intro to GAMs (Conceptual)", 
story.append(Paragraph("5. GAM Concept — Flexible Curves (10–15 min)", h1))
story.append(bullet_list([
    "[ ] Write down the GAM idea in words: ‘each feature gets its own smooth curve.’",
    "[ ] Note the additive form: prediction = baseline + f1(x1) + f2(x2) + ...",
    "[ ] List which features you’ll model with smooths (bill_total_usd, party_size, maybe gender_Male, is_smoker_Yes).",
    "[ ] In Markdown, summarize when you’d choose a GAM over plain OLS.",
]))
story.append(Spacer(1, 0.1 * inch))

# Section 6: Fit a GAM in Python
story.append(Paragraph("6. Fit a GAM in Python (20–25 min)", h1))
story.append(bullet_list([
    "[ ] Install/import pygam (or chosen GAM library).",
    "[ ] Define feature matrix X and target y (match your earlier linear model).",
    "[ ] Fit LinearGAM with one smooth term per feature (e.g., s(0) + s(1) + ...).",
    "[ ] Print/inspect model.summary() just to see overall behavior (don’t obsess over p-values).",
]))
story.append(Spacer(1, 0.1 * inch))

# Section 7: Plot & Interpret Smooth Effects
story.append(Paragraph("7. Plot & Interpret GAM Smooths (20–25 min)", h1))
story.append(bullet_list([
    "[ ] Create one plot per feature showing: x-axis = feature, y-axis = smooth effect on tip_usd.",
    "[ ] Add a horizontal line at 0 to show ‘no effect’ baseline.",
    "[ ] Save the combined ‘GAM smooths’ plot using save_and_log_plot().",
    "[ ] In Markdown, ELI5 each smooth: what story does the curve tell?",
]))
story.append(Spacer(1, 0.1 * inch))

# Section 8: Compare OLS vs GAM Performance
story.append(Paragraph("8. Compare OLS vs GAM (15–20 min)", h1))
story.append(bullet_list([
    "[ ] Compute R² and RMSE for both OLS and GAM on the same train/test split.",
    "[ ] Make a simple side-by-side table: model, R², RMSE.",
    "[ ] Write 2–3 bullet points: did GAM meaningfully improve fit?",
    "[ ] Note where GAM helps most (e.g., high bills, large parties, odd corners of the data).",
]))
story.append(Spacer(1, 0.1 * inch))

# Section 9: Diagnostics & Overfitting Check (Optional but Recommended)
story.append(Paragraph("9. Diagnostics & Overfitting Check (Optional, 10–15 min)", h1))
story.append(bullet_list([
    "[ ] Visually compare residual plots for OLS vs GAM.",
    "[ ] Check if GAM overfits: does test R² drop vs train?",
    "[ ] If overfitting appears, try adjusting GAM regularization or number of splines.",
]))
story.append(Spacer(1, 0.1 * inch))

# Section 10: Reflection & “Stickiness”
story.append(Paragraph("10. Reflection (5–10 min)", h1))
story.append(bullet_list([
    "[ ] In your own words: What can a GAM do that linear regression can’t?",
    "[ ] Write a one-sentence rule of thumb for when you’d reach for a GAM.",
    "[ ] Note 1–2 real-world problems (outside tipping) where GAMs might shine.",
]))
story.append(Spacer(1, 0.3 * inch))

# Footer note
story.append(Paragraph(
    "Tip: Next time you see curved residuals or fanning patterns, consider a GAM as a flexible, interpretable upgrade to your linear model.",
    body,
))

# Build PDF
doc.build(story)
print(f"✅ Created {OUTPUT_PATH}")
```
